```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
#State schema class
class AgentState(TypedDict):
    compliment: str  # Add a field for the generated compliment

#Node with input state and output state
def compliment_node(state: AgentState) -> AgentState:
    """
    Node that adds a compliment for a person based on the topic provided.
    """
    state['compliment'] = "Hello, you are doing great at " + state['compliment'] + " today!"

    return state
```

```python
#Build the graph to handle the state of the Agent

graph = StateGraph(AgentState)

#To add a node, you must pass the name and the action it is doing, which is your function of the node
graph.add_node("compliment", compliment_node)

#To add a start point and end point which is the node the graph should start & end at
graph.set_entry_point("compliment")
graph.set_finish_point('compliment')

#Compile the graph
app = graph.compile()
```

```python
result = app.invoke({"compliment": "coding"})

result["compliment"]
```
