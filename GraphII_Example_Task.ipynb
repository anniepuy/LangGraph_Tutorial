2. Create a processing node that performs operations on a list data
3. Set up LangGraph that processes and outputs computed results
4. Invoke the graph with structured inputs and retrieve outputs

Requirements:
If the operation is a "+" the integers will be added. If the operation is a "-" the integers will be subtracted

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph

#Create state schema with three different inputs
class AgentState(TypedDict):
    values: List[int]
    name: str
    result: str
    operation: str  # "+" or "-" as string values

```

```python
import operator 
#Create a single node that handles the three input types
def process_value(state: AgentState) -> AgentState:
    """
    This function will either add or subtract the values based on the operation string value.
    """

    if state['operation'] == "+":
        result_value = sum(state['values'])
        operation_word = "sum"
    elif state['operation'] == "-":
        result_value = state['values'][0]
        for value in state['values'][1:]:
            result_value -= value
        operation_word = "result"
                  
    state["result"] = f"Hi there {state['name']}! Your {operation_word} is {result_value}"
    return state
```

```python
#Add to graph
graph = StateGraph(AgentState)

#First variable is the name, second is the function or action
graph.add_node("processor", process_value)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

#Store the compiled graph into a variable so you can invoke using the variable
app = graph.compile()
```

```python
#invoke the graph with inputs using the variable 'app'
answers = app.invoke({"values": [1, 3, 4, 5], "name": "Lua", "operation": "-"})

print(answers)

#To just access result specify result from answer
print(answers['result'])
```
